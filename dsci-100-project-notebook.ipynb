```r
# Import library to use models
library(tidymodels)
```

```r
# Remove NA values of conf_wins_pct
# This is done because it was analyzed during checkpoint C that
# it was the only numerical variable with missing values
# and knn cannot handle them
wmm_tidy <- wmm_tidy |>
    filter(!is.na(conf_wins_pct)) # Keep the rows that are not NA

# Initial split of the data
wmm_split <- initial_split(wmm_tidy, prop = 0.75, strata = bid)
wmm_train <- training(wmm_split) # 70%
wmm_test <- testing(wmm_split)

# Recipe to preprocess data
wmm_recipe <- recipe(bid ~ seed + conf_wins_pct + reg_wins_pct, data = wmm_train) |>
    step_scale(all_predictors()) |>
    step_center(all_predictors())

# Model specification
wmm_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |>
    set_engine("kknn") |>
    set_mode("classification")
```

```r
# 5-fold cross-validation
wmm_vfold <- vfold_cv(wmm_train, v = 5, strata = bid)

# k values to try during cross-validation
k_vals = tibble(neighbors = seq(from = 2, to = 32, by = 4))

# Create the workflow
wmm_fit <- workflow() |>
    add_recipe(wmm_recipe) |>
    add_model(wmm_spec) |>
    tune_grid(resamples = wmm_vfold, grid = k_vals) |>
    collect_metrics()
```

```r
# Display the accuracy across folds for different k values
wmm_accuracies <- wmm_fit |>
    filter(.metric == "accuracy")

wmm_accuracies
```

```r
# Plot the accuracy vs. neighbors
wmm_acc_k_plot <- wmm_accuracies |>
    ggplot(aes(x = neighbors, y = mean)) +
    geom_point(size = 2) +
    geom_line() +
    labs(
        x = "Neighbors (k)",
        y = "Mean Accuracy",
        title = "Mean Accuracy Accross Neighbor Values (k)",
    ) +
    scale_x_continuous(breaks = seq(from = 2, to = 32, by = 4)) + # x-axis adjust
    scale_y_continuous(limits = c(0.73, 0.82)) + # y-axis adjust
    theme(text = element_text(size = 20))

wmm_acc_k_plot
```

```r
# Get the k value with the highest accuracy
best_k <- wmm_accuracies |>
    arrange(-mean) |> # Descendant order by mean accuracy
    head(1) |>
    pull(neighbors)

best_k
```

```r
# Model specification with best k value
wmm_best_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = best_k) |>
    set_engine("kknn") |>
    set_mode("classification")

# Fit the training data set
wmm_best_fit <- workflow() |>
    add_recipe(wmm_recipe) |>
    add_model(wmm_best_spec) |>
    fit(data = wmm_train)

# Use the final model to predict on the test data
wmm_pred <- predict(wmm_best_fit, wmm_test) |>
    bind_cols(wmm_test)

head(wmm_pred, 5)
```

```r
# Report the accuracy of the prediction
wmm_pred_accuracy <- wmm_pred |>
    metrics(truth = bid, estimate = .pred_class) |>
    filter(.metric == "accuracy") # Only report the accuracy

wmm_pred_accuracy
```
